```python
import numpy as np
```

```python
#min-max
data = np.array([10, 20, 30, 40, 50])
new_min = 0
new_max = 1
data_min = np.min(data)
data_max = np.max(data)
scaled_data = new_min + ((data - data_min) * (new_max - new_min)) / (data_max - data_min)

print("Original Data:", data)
print("Scaled Data:  ", scaled_data)
```

```python
a=np.array([10,20,30,40,50])
min_val=np.min(a)
max_val=np.max(a)
min_max = (a - min_val) / (max_val - min_val)
print(min_max)
```

```python
x = np.array([1, 2, 3, 5, 7, 8, 10, 11, 13, 15])
y = np.array([2, 4, 7, 8, 10, 15, 20, 21, 23, 30])

r_matrix = np.corrcoef(x, y)
r_value = r_matrix[0, 1]
print("Correlation Coefficient:", r_value)
```

```python
import numpy as np
from scipy import stats

observed = np.array([
    [22, 18, 10],
    [15, 25, 12]
])

print("--- Observed Frequencies ---")
print(observed)
chi2_stat, p_value, df, expected = stats.chi2_contingency(observed)

print("\n--- Expected Frequencies ---")
print(np.round(expected, 2))

print("\n--- Test Results ---")
print(f"Chi-Square Statistic : {chi2_stat:.4f}")
print(f"Degrees of Freedom   : {df}")
print(f"p-value              : {p_value:.4f}")
```
